```python
import os
import glob
import datetime
import time

import pandas as pd
import matplotlib.pyplot as plt  # Correcting matplotlib import for pyplot usage
import seaborn as sns  # Commonly alias seaborn as sns
import numpy as np  # Import numpy as np, a common alias
import sklearn  # Correcting scikit-learn import to sklearn
import scipy
import sklearn_extra  # Note: sklearn_extra is an additional library related to sklearn
import plotly.express as px  # Optional: alias plotly for easier usage
import xlsxwriter
import openpyxl
import pdfkit
```

```python
data = pd.read_csv('./champions.csv')

# Preview the first few rows
print(data.head())
# Check the structure and data types
print(data.info())
```

```python
print(data.describe())
print(data.isnull().sum())
```

```python
numerical_cols = ['Picks', 'Bans', 'Presence', 'Wins', 'Losses', 'Winrate', 'KDA', 'Avg BT', 'GT', 'CSM', 'DPM', 'GPM', 'CSD@15', 'GD@15', 'XPD@15']
for col in numerical_cols:
    sns.histplot(data[col].dropna(), kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
sns.pairplot(data[numerical_cols])
plt.show()

# Correlation heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(data[numerical_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
